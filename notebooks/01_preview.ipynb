# **Yin Yang - Parametric Visualization**

This notebook demonstrates the parametric generation of Yin Yang symbols with varying morphology. All you need is the `yinyang.py` file. You can download it from the workspace or clone the public GitHub repository.

You can generate a dataset or use the precomputed `data/sample_18k.npz` dataset. The dataset consists of an `x_train` array of images (10000, 28, 28) and a `y_train` array of the parameter **p** (10000,).

```python
# fetch repository when running on Colab
!git clone https://github.com/kwanty/YinYang.git YinYang-repo
%cd YinYang-repo

import numpy as np
import matplotlib.pyplot as plt
from yinyang import generate_yinyang, generate_dataset
```

## Single Symbol Generation

Generate a single Yin Yang symbol with parameter `p` controlling morphology. **Interpretation**:
- **p = 0.0**: Shape becomes dominated by the black lobe.
- **p = 0.5**: Balanced yin-yang form.
- **p = 1.0**: Shape becomes dominated by the white lobe.

```python
# Generate a single symbol
p = 0.5  # Morphology parameter: 0 = all black lobe, 1 = all white lobe
img = generate_yinyang(p, size=28)

plt.figure(figsize=(4, 4))
plt.imshow(img, cmap='gray')
plt.title(f'Yin Yang Symbol (p={p})')
plt.axis('off')
plt.show()
```

## 4x4 Preview Grid

Visualize the morphology progression across 16 parameter values.

```python
# Generate 4x4 grid with parameter progression
rows, cols = 4, 4
ps = np.linspace(0, 1, rows * cols)
imgs = [generate_yinyang(p, size=28) for p in ps]

fig, axes = plt.subplots(rows, cols, figsize=(10, 10))
axes = axes.flatten()

for idx, (p, ax) in enumerate(zip(ps, axes)):
    ax.imshow(imgs[idx], cmap='gray', vmin=0, vmax=1)
    ax.set_title(f'p={p:.2f}', fontsize=10)
    ax.axis('off')

plt.suptitle('Yin Yang Parametric Preview (4x4)', fontsize=14)
plt.tight_layout()
plt.show()
```

## 16x16 Full Grid

Generate a comprehensive 16x16 grid showing smooth morphology progression.

```python
# Generate 16x16 grid
rows, cols = 16, 16
total = rows * cols
ps = np.linspace(0, 1, total)
imgs = [generate_yinyang(p, size=28) for p in ps]

# Create composite image
composite = np.zeros((rows * 28, cols * 28))
for r in range(rows):
    for c in range(cols):
        idx = r * cols + c
        composite[r*28:(r+1)*28, c*28:(c+1)*28] = imgs[idx]

plt.figure(figsize=(12, 12))
plt.imshow(composite, cmap='gray', vmin=0, vmax=1)
plt.title('Yin Yang Parametric Grid (16x16)', fontsize=14)
plt.axis('off')
plt.tight_layout()
plt.show()
```
